```python
import yfinance as yf
import numpy as np
import matplotlib.pyplot as plt

apple = yf.download('AAPL', start='2020-01-01', end='2021-01-01')

apple['9_MA'] = apple['Close'].rolling(window=9).mean()
apple['20_MA'] = apple['Close'].rolling(window=20).mean()
```

```python
apple['Signal'] = np.where(apple['9_MA'] > apple['20_MA'], 1.0, 0.0)
apple['Position'] = apple['Signal'].diff()
```

```python
plt.figure()
plt.plot(apple['Close'], label='Apple Price', alpha=0.5)
plt.plot(apple['9_MA'], label='9-Day Average', alpha=0.8)
plt.plot(apple['20_MA'], label='20-Day Average', alpha=0.8)

plt.plot(apple.index,
         np.where(apple['Position'] == 1, apple['9_MA'], np.nan),
         '^', markersize=10, color='g', lw=0, label='Buy')

plt.plot(apple.index,
         np.where(apple['Position'] == -1, apple['9_MA'], np.nan),
         'v', markersize=10, color='r', lw=0, label='Sell')

plt.title('Apple Share Value')
plt.legend(loc='best')
plt.show()
```
